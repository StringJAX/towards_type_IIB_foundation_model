# Sample on GPU

After the software is installed and the kernel was restarted, you can sample from any model. The samples will be saved on your drive in the folder "IIB_foundation_model_samples" and can then be evaluated in the evaluation notebook. Evaluating on the CPU will create a new runtime and load the samples from your drive (you need to give permission). You can delete the samples from your drive once you are done. You can also directly evaluate the models here, but this will consume quite some GPU runtime.

All models from the paper can be sampled here. To speed up evaluation time, only 100 samples are drawn per prompt. If you want to recreate evaluation runs from the paper, change the number of samples to 1000. We will describe sampling from each model separately.

## 1. Sample from the pre-trained and GRPO refined model

Here you can sample from the model that has been pre-trained on all 20 geometries and further improved with GRPO. The options are the following:

- You can chose between the pre-trained and GRPO refined model by passing either "checkpoint_2000" or "GRPO_2/checkpoint_200" as load_ckpt
- You can chose a file name for the samples by changing the variable file_name
- If you want to evaluate directly on the GPU, set only_sample=False
- You can change the number of samples via the num_samples argument
- Do not change any other arguments (unless you know what you're doing)

```python
from google.colab import drive
drive.mount('/content/drive')
SAMPLE_DIR = '/content/drive/MyDrive/IIB_foundation_model_samples'
import os; os.makedirs(SAMPLE_DIR, exist_ok=True)
%cd /content/towards_type_IIB_foundation_model/transformer
from evaluate import evaluate
import pickle
samples = evaluate(RUN_ID=37, load_ckpt="GRPO_2/checkpoint_200", num_samples=100, only_sample=True)

file_name="all_h12_sample"
with open(f"{SAMPLE_DIR}/{file_name}.p", "wb") as handle:
    pickle.dump(samples, handle)
```

## 2. Sample from the fine-tuned model

Here you can sample from the model that has been pre-trained on all geometries with $h^{1,2}\leq 4$ (1.5M solutions) and then fine-tuned with a significantly smaller dataset of all geometries with $h^{1,2}=5$ (5K solutions).

- You can chose between the pre-trained and GRPO refined model by passing either "checkpoint_1000" or "GRPO_2/checkpoint_200"
- If you want to compare against the benchmark model, set RUN_ID=44 and chose between "checkpoint_1000" for the pre-trained and "GRPO_2/checkpoint_200" for the GRPO refined model
- All other settings are the same as above

```python
samples = evaluate(RUN_ID=45, load_ckpt="GRPO_2/checkpoint_200", num_samples=100, only_sample=True)

file_name="h12_5_finetune_sample"
with open(f"{SAMPLE_DIR}/{file_name}.p", "wb") as handle:
    pickle.dump(samples, handle)
```

## 3. Sample conditioned on $N_{\rm flux}, |W_0|$ or $1/g_s$

Here you can sample from the models that have additional been conditioned on $N_{\rm flux}, |W_0|$ or $1/g_s$. 

### $N_{\rm flux}$

- You can chose between the pre-trained and GRPO refined model by passing either "checkpoint_2000" or "GRPO_8/checkpoint_400"
- You can chose a target value for $N_{\rm flux}$ by passing a different nflux (has to be an integer!) to the evaluation function
- All other settings are the same as above

```python
samples = evaluate(RUN_ID=38, load_ckpt="GRPO_8/checkpoint_400", num_samples=100, only_sample=True, nflux=80)

file_name="n_flux_sample"
with open(f"{SAMPLE_DIR}/{file_name}.p", "wb") as handle:
    pickle.dump(samples, handle)
```

### $|W_0|$

- You can chose between the pre-trained and GRPO refined model by passing either "checkpoint_2000" or "GRPO_2/checkpoint_400"
- You can chose a target value for $|W_0|$ by passing a different w0 to the evaluation function
- All other settings are the same as above

```python
samples = evaluate(RUN_ID=39, load_ckpt="GRPO_2/checkpoint_400", num_samples=100, only_sample=True, w0=1.0)

file_name="w0_sample"
with open(f"{SAMPLE_DIR}/{file_name}.p", "wb") as handle:
    pickle.dump(samples, handle)
```

### $1/g_s$

- You can chose between the pre-trained and GRPO refined model by passing either "checkpoint_2000" or "GRPO_5/checkpoint_400"
- You can chose a target value for $1/g_s$ by passing a different gs_inv to the evaluation function
- All other settings are the same as above

```python
samples = evaluate(RUN_ID=46, load_ckpt="GRPO_5/checkpoint_400", num_samples=100, only_sample=True, gs_inv=3.0)

file_name="gs_inv_sample"
with (f"{SAMPLE_DIR}/{file_name}.p").open("wb") as handle:
    pickle.dump(samples, handle)
```
